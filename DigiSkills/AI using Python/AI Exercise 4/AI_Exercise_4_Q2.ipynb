```python
import json

with open("exercise_text.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print("JSON file loaded successfully!")
print("\nFile content:")
print(data)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

print("Hugging Face tokenizer loaded successfully!")
```

```python
# Extract text from JSON file
text = data["text"]

# Tokenize the text
tokens = tokenizer(
    text,
    padding=True,
    truncation=True,
    return_tensors="np"
)

# Display results
print("Tokenization completed successfully!")

print("\nOriginal Text:")
print(text)

print("\nTokens:")
print(tokenizer.convert_ids_to_tokens(
    tokens["input_ids"][0].tolist()
))

print("\nToken IDs:")
print(tokens["input_ids"][0])

print("\nAttention Mask:")
print(tokens["attention_mask"][0])

print("\nTotal Tokens:", len(tokens["input_ids"][0]))
```
